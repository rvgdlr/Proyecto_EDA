## Importar librerías Panda
- Visualizar todas las columnas.

```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

## Descargar archivo CSV

```python
df = pd.read_csv('../data/bank-additional.csv')
```

## Visulaizar aleatoriamente 5 filas para ver una muestra de los datos.

```python
df.sample(5)
```

## Comprobar tamaño del conjunt de datos.

```python
print(f'Nuestro conjunto de datos presenta un total de {df.shape[0]} filas y {df.shape[1]} columnas')
```

## Analizar los nulos y el tipo de dato.

```python
df.info()
```

## Ver porcentaje de nulos sólo en las columnas donde hayan nulos. Redondear a dos decimales.

```python
nulos_porcentaje = df.isnull().mean()*100
nulos_porcentaje[nulos_porcentaje>0].round(2)
```

## Comprobarmos que no hayan duplicidades.

```python
df.duplicated().sum()
```

## Analizamos las columnas categóricas más detenidamente, para comprobar si hay algo más que limpiar.

```python
for col in df.select_dtypes(include ='object').columns:
    print(col.upper())
    print(df[col].value_counts())
    print('--------')
```
