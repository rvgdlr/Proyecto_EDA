# Metricas a estudiar
- cons.conf.idx: indice de confianza del consumidor por cada empleado
- previous: número de veces que se ha contactado con el cientes antes de esta campaña por cada empleado
- duration: duración en segundos de la última interación de cada cliente por empleados
- pdays: número de días que han pasado desde la última vez que se contactó con el cliente

```python
fig, axes = plt.subplots(nrows =1, ncols= 4, figsize=(25,5))

sns.barplot(x = 'nr_employed', y = 'cons_conf_idx', data = df, ax = axes[0], hue= 'nr_employed', palette='cool')
axes[0].set_title('cons_conf_idx')
sns.barplot(x = 'nr_employed', y = 'previous', data = df, ax = axes[1], hue= 'nr_employed', palette='cool')
axes[0].set_title('previous')
sns.barplot(x = 'nr_employed', y = 'duration', data = df, ax = axes[2], hue= 'nr_employed', palette='cool')
axes[0].set_title('duration')
sns.barplot(x = 'nr_employed', y = 'pdays', data = df, ax = axes[3], hue= 'nr_employed', palette='cool')
axes[0].set_title('pdays')
```
