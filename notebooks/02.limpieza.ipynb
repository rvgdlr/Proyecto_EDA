- ** job ** -> Tiene 345 nulos. CAmbiar admin. por administrative

```python
# Tratamiento de datos
import pandas as pd
pd.set_option('display.max_columns', None)
#----------------
#Exploración de archivos
import sys
sys.path.append('../')
#------------------
#Funciones personalizadas
from src import sp_limpieza as lim
```

```python
df_raw = pd.read_csv('../data/bank-additional.csv')
df_raw.sample(5)
```

```python
df = df_raw.copy()
```

### Cambios nombres columnas

```python
df.columns = df.columns.str.replace('.','_')
```

```python
df.columns
```

### date cambiar a formato fecha

```python
meses = {
    'enero': 'January',
    'febrero': 'February',
    'marzo': 'March',
    'abril': 'April',
    'mayo': 'May',
    'junio': 'June',
    'julio': 'July',
    'agosto': 'August',
    'septiembre': 'September',
    'octubre': 'October',
    'noviembre': 'November',
    'diciembre': 'December'
}

df['date'] = df['date'].str.lower().replace(meses, regex=True)
df['date'] = pd.to_datetime(df['date'], format='%d-%B-%Y')
```

```python
df.info()
```

### marital-> Esta en mayúsculas cambiar a minúsculas

```python
df['marital'] = df['marital'].str.lower()
```

### job-> Cambiar admin. por administrative

```python
df['job'] = df['job'].str.replace('admin.', 'administrative')
```

### eduction -> Hay puntos separando las palabras cambiar '.' por '_'. 

```python
df['education']= df['education'].str.replace('.', '_')
```

### cons.price.idx -> Cambiar cambiar ',' por '.' y pasar a tipo float
### cons.conf.idx -> Cambiar cambiar ',' por '.' y pasar a tipo float
### euribor3m -> Cambiar cambiar ',' por '.' y pasar a tipo float


```python
lim.comas(df,'cons_conf_idx')

```

### Age -> El número es float, para la edad del cliente no es necesario

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df['age'] = df['age'].astype('Int64')
```

###  nr_employed -> Cambiar a tipo integer quitando las ','.

```python
df['nr_employed'] = pd.to_numeric(df['nr_employed'], errors='coerce')
```

```python
df['nr_employed'] = df['nr_employed'].round(0).astype('Int64')
```

### default -> Los valores tienen que ser 1 o 0, por lo que podemos cambiarlo de float a integer.
### housing -> Los valores tienen que ser 1 o 0, por lo que podemos cambiarlo de float a integer.
### loan -> Los valores tienen que ser 1 o 0, por lo que podemos cambiarlo de float a integer.

```python
lim.enteros (df, 'default')
lim.enteros (df, 'housing')
lim.enteros (df, 'loan')
```
